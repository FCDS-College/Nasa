```python
import json
import os

# المسارات الأساسية للنظام
TEST_FILE = "datasets/datasets/evaluation/rag_gold.jsonl" # مسار ملف الاختبارات
DOCS_DIR = "datasets/documents/"     # مسار مجلد المستندات
MEDIA_DIR = "datasets/media/"       # مسار مجلد الوسائط

def verify_and_run_system():
    print("="*60)
    print(" 🎯 قالب التشغيل السريع لداتا الفريق (Team Data Integration & Test)")
    print("="*60)

    # 1. التحقق من وجود ملف الاختبارات الرئيسي
    if not os.path.exists(TEST_FILE):
        print(f"❌ تنبيه: ملف الاختبارات غير موجود في مساره: {TEST_FILE}")
        print("💡 اطلب من زميلك المسؤول عن الـ JSONL تسليم الملف ووضعه في مجلد datasets/")
        return

    # 2. قراءة حالات الاختبار والتحقق من الملفات المرتبطة بها (نصوص ووسائط)
    total_cases = 0
    missing_media_count = 0
    
    print("\n🔍 جارٍ فحص حالات الاختبار والملفات المرتبطة...")
    with open(TEST_FILE, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            
            case = json.loads(line)
            total_cases += 1
            test_id = case.get("test_id", "Unknown")
            visual_input = case.get("visual_input")

            # لو الحالة فيها مدخل بصري، نتاكد إنه موجود في مجلد الـ media
            if visual_input:
                if not os.path.exists(visual_input):
                    print(f"  ⚠️ تحذير في الاختبار [{test_id}]: الملف البصري المطلوب غير موجود -> {visual_input}")
                    missing_media_count += 1

    print(f"\n📊 إحصائيات سريعة قبل التشغيل:")
    print(f"   - إجمالي حالات الاختبار في الملف: {total_cases}")
    print(f"   - الملفات البصرية الناقصة: {missing_media_count}")

    if missing_media_count > 0:
        print("\n💡 نصيحة: تأكد من أن زملاءك قاموا بوضع الصور/الفيديوهات في المجلدات المخصصة قبل البدء الفعلي.")
    else:
        print("\n✨ كل الملفات البصرية سليمة وموجودة في أماكنها الصحيحة!")

    print("\n🚀 النظام جاهز تماماً! يمكنك الآن تشغيل Master Evaluator أو pytest للبدء الفعلي.")
    print("="*60)

if __name__ == "__main__":
    verify_and_run_system()
```
