### Adversarial Robustness Testing

```python
import json
import os

HISTORY_FILE = "experiment_history.json"

class PerformanceDriftMonitor:
    def __init__(self):
        print("📈 تم تفعيل نظام مراقبة انحراف وتطور الأداء (Performance Drift Monitor)...")

    def analyze_drift(self):
        """
        مقارنة أداء التجارب السابقة لاكتشاف ما إذا كان أداء الموديل 
        يتحسن أم يتراجع بمرور الوقت مع التحديثات.
        """
        if not os.path.exists(HISTORY_FILE):
            print("⚠️ لا يوجد سجل تجارب سابق للمقارنة.")
            return

        with open(HISTORY_FILE, 'r', encoding='utf-8') as f:
            try:
                history = json.load(f)
            except json.JSONDecodeError:
                history = []

        if len(history) < 2:
            print("📊 عدد التجارب مسجل كبيانات أولية، يلزم تجربتان على الأقل لمقارنة الانحراف.")
            return

        latest = history[-1]["metrics"]
        previous = history[-2]["metrics"]

        print(f"🔬 مقارنة التجربة الأخيرة (#{history[-1]['experiment_id']}) بالتجربة السابقة (#{history[-2]['experiment_id']}):")
        
        for key in latest:
            if key in previous:
                diff = latest[key] - previous[key]
                trend = "🟢 تحسن" if diff > 0 else ("🔴 تراجع" if diff < 0 else "⚪ ثبات")
                print(f"  - {key}: السابق ({previous[key]}) ➔ الحالي ({latest[key]}) | النتيجة: {trend} (فرق: {diff:+.2f})")

if __name__ == "__main__":
    monitor = PerformanceDriftMonitor()
    monitor.analyze_drift()
```
