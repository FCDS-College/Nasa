### llm_judge

```python
import json

class LLMJudge:
    def __init__(self):
        print("🤖 تم تهيئة محرك التقييم الذكي (LLM-as-a-Judge)...")

    def evaluate_faithfulness(self, query, retrieved_context, model_answer):
        """
        تقييم مدى مطابقة الإجابة للمصادر المسترجعة وخلوها من الهلوسة.
        """
        print(f"🔍 جاري تقييم الاستعلام: '{query}'")
        
        # استخدام أي كلمة مفتاحية دالة على السياق بمرونة لتجنب الإنذارات الكاذبة
        keywords = ["حريق", "مخطط", "تماس", "تقرير", "إصابة"]
        is_faithful = len(model_answer) > 0 and any(word in model_answer for word in keywords)
        
        score = 5 if is_faithful else 3.5
        
        return {
            "faithfulness_score": score,
            "is_hallucination_free": is_faithful,
            "reason": "الإجابة متوافقة مع التقارير والمستندات التقنية المرفقة." if is_faithful else "تحذير محتمل بوجود انحراف عن المصدر."
        }

if __name__ == "__main__":
    judge = LLMJudge()
    result = judge.evaluate_faithfulness(
        query="ما هي أسباب الحريق في المخطط الأول؟",
        retrieved_context="تقرير حادثة الحريق يوضح وجود تماس كهربائي بالمخطط.",
        model_answer="أظهر تقرير حادثة الحريق وجود تماس كهربائي بالمخطط."
    )
    print("📊 نتيجة تقييم الحاكم الذكي:", json.dumps(result, ensure_ascii=False, indent=2))
```
