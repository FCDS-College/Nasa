```python
import json
import os
from datetime import datetime

HISTORY_FILE = "experiment_history.json"

def log_experiment(metrics_dict):
    """تسجيل نتائج التجربة الحالية في سجل الأداء التاريخي"""
    history = []
    if os.path.exists(HISTORY_FILE):
        with open(HISTORY_FILE, 'r', encoding='utf-8') as f:
            try:
                history = json.load(f)
            except json.JSONDecodeError:
                history = []

    experiment_record = {
        "experiment_id": len(history) + 1,
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "metrics": metrics_dict
    }
    
    history.append(experiment_record)

    with open(HISTORY_FILE, 'w', encoding='utf-8') as f:
        json.dump(history, f, ensure_ascii=False, indent=4)
    
    print(f"📈 تمت إضافة التجربة رقم #{experiment_record['experiment_id']} بنجاح إلى سجل التاريخ!")

if __name__ == "__main__":
    sample_metrics = {
        "accuracy": 0.95,
        "f1_score": 0.92,
        "faithfulness": 4.8
    }
    log_experiment(sample_metrics)
```
