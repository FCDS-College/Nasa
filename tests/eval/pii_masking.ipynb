### Data Anonymization / PII Masking

```python
import re
import json

class PIIMaskingGuardrail:
    def __init__(self):
        print("🔒 تم تفعيل طبقة تعمية البيانات الحساسة (PII Masking Guardrail)...")

    def mask_sensitive_data(self, text):
        """
        إخفاء أو تعمية البيانات الحساسة مثل أرقام الهواتف، البريد الإلكتروني، 
        أو الأسماء السرية في تقارير الحرائق لضمان الخصوصية والأمان.
        """
        print("🔍 جاري فحص النص وتعمية البيانات الحساسة...")
        
        # تعمية البريد الإلكتروني
        masked_text = re.sub(r'[\w\.-]+@[\w\.-]+\.\w+', '[EMAIL_REDACTED]', text)
        
        # تعمية أرقام الهواتف (كمثال مبسط للأرقام الطويلة)
        masked_text = re.sub(r'\b01[0-9]{9}\b', '[PHONE_REDACTED]', masked_text)
        
        # تعمية الأكواد أو المعرفات السرية لو وجدت
        masked_text = re.sub(r'REF-\d+', '[CONFIDENTIAL_REF]', masked_text)

        return {
            "original_preview": text[:30] + "...",
            "masked_output": masked_text,
            "status": "SECURED"
        }

if __name__ == "__main__":
    guard = PIIMaskingGuardrail()
    sample_report = "تقرير حريق المبنى رقم REF-402، المسؤول عن البلاغ أحمد عبر البريد ahmed@fire-dept.com أو هاتف 01012345678."
    result = guard.mask_sensitive_data(sample_report)
    print("📊 نتيجة تأمين البيانات:", json.dumps(result, ensure_ascii=False, indent=2))
```
