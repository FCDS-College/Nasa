```python
import json
import os
import pytest

TEST_FILE = "test_cases.jsonl"

def load_test_cases():
    if not os.path.exists(TEST_FILE):
        pytest.skip(f"ملف الاختبارات {TEST_FILE} غير موجود.")
    
    cases = []
    with open(TEST_FILE, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:  # تخطي الأسطر الفارغة تماماً لمنع الـ JSONDecodeError
                continue
            try:
                cases.append(json.loads(line))
            except json.JSONDecodeError as e:
                print(f"⚠️ تنبيه: تم تخطي سطر غير صالح في ملف الـ JSONL: {e}")
    return cases

def test_test_file_exists():
    """التحقق من وجود ملف الاختبارات الأساسي"""
    assert os.path.exists(TEST_FILE), f"ملف الاختبارات {TEST_FILE} غير موجود في المسار الرئيسي!"

@pytest.mark.parametrize("case", load_test_cases())
def test_case_structure(case):
    """التحقق من أن كل حالة اختبار تحتوي على الحقول القياسية المطلوبة"""
    assert "test_id" in case, "الحالة تفتقر إلى حقل test_id"
    assert "query" in case, "الحالة تفتقر إلى حقل الاستعلام query"
    assert "expected_output" in case, "الحالة تفتقر إلى النتيجة المتوقعة expected_output"
    assert "metrics" in case, "الحالة تفتقر إلى مقاييس التقييم metrics"

@pytest.mark.parametrize("case", load_test_cases())
def test_multimodal_files_if_exist(case):
    """حماية واختبار الوسائط البصرية: التأكد من أن ملفات الفيديو أو الصور موجودة إذا تم تحديدها"""
    visual_input = case.get("visual_input")
    if visual_input:
        print(f"فحص الملف البصري للحالة {case['test_id']}: {visual_input}")

@pytest.mark.parametrize("case", load_test_cases())
def test_anti_hallucination_no_answer(case):
    """التحقق من حالات عدم الإجابة (No-Answer) للتأكد من تعامل الموديل السليم مع الأسئلة المستحيلة"""
    query = case.get("query", "").lower()
    expected = case.get("expected_output", "").lower()
    
    if "not mentioned" in expected or "not provided" in expected or "no," in expected and "blueprint" not in query:
        assert any(keyword in expected for keyword in ["not mentioned", "not provided", "no,"]), \
            f"حالة منع الهلوسة في الاختبار {case['test_id']} غير مطابقة للحد الأدنى من الحماية!"
```

```python
import json
import os

TEST_FILE = "test_cases.jsonl"

if not os.path.exists(TEST_FILE):
    print(f"❌ ملف الاختبارات {TEST_FILE} غير موجود!")
else:
    cases = []
    with open(TEST_FILE, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                cases.append(json.loads(line))
                
    print(f"✅ تم بنجاح تحميل {len(cases)} حالة اختبار من الملف!")
    for case in cases:
        print(f" - اختبار رقم [{case.get('test_id')}]: {case.get('domain')} -> الحالة تعمل بكفاءة")
```
