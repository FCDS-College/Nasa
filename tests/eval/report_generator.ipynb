### report_generator

```python
import json
import os
from datetime import datetime

def generate_evaluation_report(results_data, output_path="evaluation_report.json"):
    """توليد وحفظ تقرير التقييم النهائي كملف بيانات إنتاجي"""
    report = {
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "project": "Flame in Freefall - AI Evaluation Framework",
        "status": "SUCCESS",
        "metrics_summary": results_data
    }
    
    with open(output_path, 'w', encoding='utf-8') as f:
        json.dump(report, f, ensure_ascii=False, indent=4)
    
    print(f"📁 تم بنجاح حفظ تقرير التقييم النهائي في المسار: {output_path}")

if __name__ == "__main__":
    # عينة بيانات افتراضية من نتائج التقييم اللي طلعناها
    sample_results = {
        "total_tests_processed": 6,
        "passed_protection_layer": 6,
        "successful_alignments": 3,
        "failed_or_skipped": 0,
        "average_retrieval_relevance": 4.5,
        "average_faithfulness": 5.0
    }
    generate_evaluation_report(sample_results)
```
