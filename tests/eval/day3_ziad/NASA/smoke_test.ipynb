### Import 

```python
import json
import os
```

### run_smoke_test

```python
def run_smoke_test(file_path):
    print("🔍 Checking contracts and starting the smoke test...")
    
    if not os.path.exists(file_path):
        print(f"❌ Error: File not found: {file_path}")
        return False
        
    # Required fields in each test case.
    required_keys = {"test_id", "domain", "query", "expected_output", "expected_citations", "metrics"}
    metric_keys = {"retrieval_relevance", "faithfulness", "citation_accuracy"}
    
    line_count = 0
    with open(file_path, 'r', encoding='utf-8') as f:
        for i, line in enumerate(f, 1):
            line_count += 1
            try:
                data = json.loads(line)
            except json.JSONDecodeError:
                print(f"❌ Error on line {i}: Invalid JSON format.")
                return False
            
            # Check for required top-level fields.
            missing_keys = required_keys - data.keys()
            if missing_keys:
                print(f"❌ Error on line {i} (test {data.get('test_id', 'unknown')}): Missing fields: {missing_keys}")
                return False
            
            # Check for required metric fields.
            metrics = data.get("metrics", {})
            missing_metrics = metric_keys - metrics.keys()
            if missing_metrics:
                print(f"❌ Error on line {i}: Missing evaluation metrics: {missing_metrics}")
                return False
                
    print(f"✅ Smoke and contract tests passed! Successfully validated {line_count} test cases.")
    return True

if __name__ == "__main__":
    run_smoke_test('test_cases.jsonl')
```
