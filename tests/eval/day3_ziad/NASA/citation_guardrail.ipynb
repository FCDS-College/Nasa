### Citation Verification & Grounding Guardrail

### citation_guardrail

```python
import json

class CitationVerificationGuardrail:
    def __init__(self):
        print("🛡️ تم تفعيل طبقة التحقق من الاستشهادات والمصادر (Citation Guardrail)...")

    def verify_response_citations(self, model_answer, retrieved_context, expected_citations):
        """
        التحقق من أن الإجابة تحتوي على استشهادات صحيحة ومستمدة من المصادر المتاحة 
        وأنها غير مبنية على هلوسة وهمية.
        """
        print("🔍 جاري فحص ومطابقة الاستشهادات للأحداث والمخططات...")
        
        # 1. فحص هل الإجابة تحتوي على مؤشرات مرجعية أو مصادر واضحة
        has_citations_in_text = any(cit in model_answer for cit in expected_citations) if expected_citations else True
        
        # 2. فحص ما إذا كانت الإجابة تعتمد على السياق المسترجع فعلياً
        is_grounded = len(model_answer.strip()) > 0 and (has_citations_in_text or "مخطط" in model_answer or "تقرير" in model_answer)

        if not is_grounded or not has_citations_in_text:
            return {
                "status": "BLOCKED",
                "safe_to_publish": False,
                "reason": "❌ تم حظر الإجابة: الموديل حاول تقديم معلومات غير موثقة بمصادر أو استشهادات صحيحة (هلوسة)."
            }
        
        return {
            "status": "APPROVED",
            "safe_to_publish": True,
            "reason": "✨ اجتازت الإجابة فحص الاستشهادات والمصادر بنجاح وهي موثقة بالكامل."
        }

if __name__ == "__main__":
    guardrail = CitationVerificationGuardrail()
    
    # تجربة لحالة ناجحة وموثقة
    test_result = guardrail.verify_response_citations(
        model_answer="أظهر تقرير الحريق رقم [REF-101] وجود تماس كهربائي.",
        retrieved_context="تقرير حادثة الحريق رقم [REF-101] يوضح تفاصيل التماس الكهربائي.",
        expected_citations=["[REF-101]"]
    )
    
    print("📊 نتيجة فحص الحماية:", json.dumps(test_result, ensure_ascii=False, indent=2))
```
