### XAI - Explainable AI Report Generator

```python
import json
from datetime import datetime

class XAIReporter:
    def __init__(self):
        print("📊✨ تم تفعيل نظام تقارير الشفافية والتفسيرية (XAI Reporter)...")

    def generate_explanation_report(self, test_id, decision, factors):
        """
        توليد تقرير تفصيلي يشرح سبب اتخاذ الموديل لقرار معين أو سبب حظر الإجابة
        للقضاء نهائياً على مشكلة 'الصندوق الأسود' للذكاء الاصطناعي.
        """
        print(f"📝 جاري صياغة تقرير التفسير والشفافية للحالة [{test_id}]...")
        
        report = {
            "report_id": f"XAI-{test_id}-{int(datetime.now().timestamp())}",
            "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "target_test_id": test_id,
            "final_model_decision": decision,
            "explanation_factors": factors,
            "transparency_note": "هذا التقرير مولد أوتوماتيكياً لتوضيح وزن كل ميزة في اتخاذ القرار النهائي."
        }
        
        return report

if __name__ == "__main__":
    reporter = XAIReporter()
    sample_report = reporter.generate_explanation_report(
        test_id="TC_042",
        decision="BLOCKED_BY_GUARDRAIL",
        factors={
            "retrieval_relevance_score": 2.1,
            "citation_match": False,
            "visual_consistency": True,
            "primary_reason": "نقص الأدلة الموثقة وغياب المراجع الدقيقة."
        }
    )
    print("📋 تقرير الشفافية المُتولد:", json.dumps(sample_report, ensure_ascii=False, indent=2))
```
